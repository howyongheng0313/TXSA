```python
import re

# Read file
with open("DATASET/Data_1.txt", "r", encoding="utf-8") as file:
    text = file.read()

words = re.findall(r'\b\w+\b', text.lower())
seen = set()
unique_words = [w for w in words if not (w in seen or seen.add(w))]

def regex_stem(word):
    # Matches the core stem in group 1, and common suffixes in group 2
    pattern = r'^(.*?)(ing|ly|ed|ious|ies|ive|es|s|ment|ational|tional|al|ion)$'
    return re.sub(pattern, r'\1', word)

# Apply the regex stemmer
stemmed_words = [regex_stem(w) for w in unique_words]
print("Regex Stemmer Output:")
print(stemmed_words)
```

```python
import re
from nltk.stem import PorterStemmer

with open("DATASET/Data_1.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Extract unique words
words = re.findall(r'\b\w+\b', text.lower())
seen = set()
unique_words = [w for w in words if not (w in seen or seen.add(w))]

# Initialize 
porter = PorterStemmer()

# Apply the stemmer
stemmed_words = [porter.stem(w) for w in unique_words]
print("Porter Stemmer Output:")
print(stemmed_words)
```

```python
import re
from nltk.stem import LancasterStemmer

with open("DATASET/Data_1.txt", "r", encoding="utf-8") as file:
    text = file.read()

words = re.findall(r'\b\w+\b', text.lower())
seen = set()
unique_words = [w for w in words if not (w in seen or seen.add(w))]

lancaster = LancasterStemmer()
stemmed_words = [lancaster.stem(w) for w in unique_words]
print("Lancaster Stemmer Output:")
print(stemmed_words)
```
