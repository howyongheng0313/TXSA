##### **1) Word Tokenization using RegEx, NLTK and SpaCY**

```python
# RegEx
import re

# 1) Match consecutive word characters or
# 2) Match one character that is neither word character or whitespace
regex_tokens = re.findall(r"\w+|[^\w\s]", text)

print("RegEx tokens:")
print(regex_tokens)
print("\nTotal tokens:", len(regex_tokens))
```

```python
# NLTK
import nltk
from nltk.tokenize import word_tokenize

# A data packages that provides pretrained information to help sentence splitting
nltk.download('punkt_tab')

nltk_tokens = word_tokenize(text)

print("NLTK tokens:")
print(nltk_tokens)
print("\nTotal tokens:", len(nltk_tokens))
```

```python
# SpaCY
import spacy

# Create English tokeniser using built-in language rules
nlp = spacy.blank("en")

# Tokenise the text and store as Doc object
doc = nlp(text)

# Extract token text, exclude whitespace only token
spacy_tokens = []
for token in doc:
    if not token.is_space:
        spacy_tokens.append(token.text)

print("spaCy tokens:")
print(spacy_tokens)
print("\nTotal tokens:", len(spacy_tokens))
```

##### **3) Stop word and Punctuation Removal**

```python
import nltk
import string
from nltk.corpus import stopwords

# Download and load the English stop-word list.
nltk.download("stopwords", quiet=True)
stop_words = set(stopwords.words("english"))

# Empty list
stop_words_found = []
punctuation_found = []
filtered_tokens = []

for token in nltk_tokens:
    # Compare in lowercase so "The" and "the" are both recognised.
    if token.lower() in stop_words:
        stop_words_found.append(token.lower())

    # Remove tokens made entirely of punctuation.
    elif token in string.punctuation:
            punctuation_found.append(token)

    # Keep the remaining tokens in their original order and case.
    else:
        filtered_tokens.append(token)

# Show each stop word and punctuation symbol found only once.
print("Unique stop words found:")
print(sorted(set(stop_words_found)))

print("\nUnique punctuation symbols found:")
print(sorted(set(punctuation_found)))

print("\nTokens after stop-word and punctuation removal:")
print(filtered_tokens)

# Counts include repeated occurrences.
print("\nOriginal token count:", len(nltk_tokens))
print("Stop-word tokens removed:", len(stop_words_found))
print("Punctuation tokens removed:", len(punctuation_found))
print("Final token count:", len(filtered_tokens))
```
